```python
from math import comb
import matplotlib.pyplot as plt
```

```python
def check_prob(n):
    return comb(365,n)/(365**n)
```

```python
n = [i for i in range(1,100)]
```

```python
probs = []
for i in n:
    prob = check_prob(i)
    probs.append(prob)
```

```python
plt.figure(figsize=(10,4))
plt.plot(n,probs,)
plt.title('Atleast Two share same birthday')
plt.xlabel("Students (n)")
plt.ylabel('Probabilites P(x)')
plt.show()
```
